Formally derive why the raw Maximum Likelihood Estimator for the covariance matrix, $\hat{\Sigma}_{MLE}$, is inherently biased in finite samples, by showing that:$$\mathbb{E}[\hat{\Sigma}_{MLE}]=\frac{n-1}{n}\Sigma$$Proof:We begin with the Maximum Likelihood Estimator (MLE) for the population covariance matrix $\Sigma$:

To evaluate the expectation of this estimator and check for bias, we expand the sample deviations around the true, hidden population mean $\mu$:
  $$X_{i}-\hat{\mu}_{n} = (X_{i}-\mu) - (\hat{\mu}_{n}-\mu)$$Substituting this identity back into the summation of the MLE formula yields:
  $$\sum_{i=1}^{n}(X_{i}-\hat{\mu}_{n})(X_{i}-\hat{\mu}_{n})^{T} = \sum_{i=1}^{n}\left[(X_{i}-\mu)-(\hat{\mu}_{n}-\mu)\right]\left[(X_{i}-\mu)-(\hat{\mu}_{n}-\mu)\right]^{T}$$Expanding the product matrix term-by-term:

This collapses the two middle cross-product terms and the final term as follows:

The Lindeberg-Lévy Multivariate Central Limit Theorem establishes that as the sample snapshot size approaches infinity ($n \rightarrow \infty$), the standardized difference between the empirical sample mean and the population mean converges in distribution to a Multivariate Gaussian distribution:

  $$\sqrt{n}(\hat{\mu}_{n}-\mu)\xrightarrow{d}\mathcal{N}(0,\Sigma)$$However, the true population covariance matrix $\Sigma$ is a latent, unobservable parameter. To handle this in practice, we rely on Slutsky's Theorem to substitute our empirical sample covariance matrix $S$.  1. Rigorous Statement of Slutsky's TheoremLet $\{X_{n}\}$ be a sequence of random vectors and $\{Y_{n}\}$ be a sequence of random matrices (or scalar random variables).
If $X_{n}$ converges in distribution to a random variable/vector $X$ ($X_{n}\xrightarrow{d}X$) and $Y_{n}$ converges in probability to a constant matrix or scalar $c$ ($Y_{n}\xrightarrow{p}c$), then:  Summation Rule:$$X_{n} + Y_{n} \xrightarrow{d} X + c$$  Multiplication Product Rule:$$X_{n}Y_{n} \xrightarrow{d} cX \quad \text{(or } Y_{n}X_{n} \xrightarrow{d} cX\text{)}$$  Division/Inverse Rule:$$X_{n}/Y_{n} \xrightarrow{d} X/c \quad \text{or} \quad Y_{n}^{-1}X_{n} \xrightarrow{d} c^{-1}X \quad (\text{provided } c \neq 0)$$  2. Operational Distribution Justification ProofStep-by-step Mathematical Proof:Step 1: Normalize the Gaussian Space By the Lindeberg-Lévy Multivariate CLT, the un-normalized mean variance sequence limits to:
  $$\sqrt{n}(\hat{\mu}_{n}-\mu)\xrightarrow{d}\mathcal{N}(0,\Sigma)$$If we multiply this expression by the inverse matrix square root of the true, latent population covariance matrix ($\Sigma^{-1/2}$), we structurally map the distribution into an isotropic, standard Gaussian space:
  $$\Sigma^{-1/2}\left[\sqrt{n}(\hat{\mu}_{n}-\mu)\right]\xrightarrow{d}\mathcal{N}(0,I)$$  Step 2: Establish the Convergence of the Estimator By applying the Weak Law of Large Numbers (WLLN), the unbiased empirical sample covariance matrix $S$ acts as a consistent estimator of the population matrix $\Sigma$. This means that as the finite snapshot history approaches infinity ($n \rightarrow \infty$), $S$ collapses directly to its population constant:
  $$S\xrightarrow{p}\Sigma$$  Since the inverse matrix square root is a continuous function mapping over positive-definite matrices, applying the continuous mapping theorem preserves this convergence:
  $$S^{-1/2}\xrightarrow{p}\Sigma^{-1/2}$$  Step 3: Combine Layers via Slutsky's Theorem We can now apply Slutsky's Multiplication Product Rule. We multiply our sequence of random vectors (which converges in distribution) by our sequence of empirical sample matrices (which converges in probability to a constant matrix):
  $$S^{-1/2}\left[\sqrt{n}(\hat{\mu}_{n}-\mu)\right] \xrightarrow{d} \Sigma^{-1/2} \cdot \mathcal{N}(0,\Sigma) = \mathcal{N}(0,I)$$  Step 4: Scale Back to the Operational Metric Because the underlying tracking space converges cleanly to a standard white-noise Gaussian limit $\mathcal{N}(0,I)$, we can map the localized coordinate parameters back to the empirical matrix metric:
  $$\sqrt{n}(\hat{\mu}_{n}-\mu)\sim\mathcal{N}(0,S)$$  Dividing across by the sample scale factor $\sqrt{n}$ yields our final operational distribution used in real-time predictive dashboards:
  $$\hat{\mu}_{n}\sim\mathcal{N}\left(\mu,\frac{1}{n}S\right)$$
  
  
   Conclusion: This proof mathematically validates that replacing the unobservable matrix $\Sigma$ with the empirical matrix $S$ does not introduce downstream structural distortion or compromise the asymptotic stability of the system's operational diagnostic envelope

1.**Mathematical Tracking Framework Setup**

To verify data stationarity and test for Constant Joint Mean (First Moment Homogeneity) across the time index , we partition our dataset of $n=5,000$ snapshots across $m=3$ channels into $g=5$ equal, consecutive, non-overlapping blocks of size $n_j = 1,000$.

  We carry out a Multivariate Analysis of Variance (MANOVA) from scratch using the following definitions:

  O. **Global Mean Vector ($\bar{X}$)**: The overall mean across all samples.

  O. **Chunk Mean Vector ($\bar{X}_j$)**: The mean vector computed within the $j$-th chunk ($j = 1, \dots, g$).

  O. **Within-Chunk Sum of Squares and Cross-Products Matrix ($W$)**: Measures internal random variations inside the partitioned intervals:

  $$W = \sum_{j=1}^{g} \sum_{i \in \text{chunk } j} (X_{ij} - \bar{X}_j)(X_{ij} - \bar{X}_j)^T$$Between-Chunk Sum of Squares and Cross-Products Matrix ($B$): Measures the spatial variance between different chunk centroids:
  $$B = \sum_{j=1}^{g} n_j (\bar{X}_j - \bar{X})(\bar{X}_j - \bar{X})^T$$Using these internal energy matrices, we calculate Wilks' Lambda ($\Lambda$):
  $$\Lambda = \frac{|W|}{|B + W|}$$To determine statistical significance, we transform the determinant ratio into an asymptotic chi-square metric using Bartlett's $\chi^2$ approximation:
  $$\chi_{calc}^{2} = -\left(n - 1 - \frac{m + g}{2}\right) \ln \Lambda$$The degrees of freedom for evaluating the resulting p-value correspond to $df = m(g - 1) = 3 \times (5 - 1) = 12$.  2. Complete Python Implementation (From Scratch)Below is the clean, self-contained Python program to reproduce the synthetic data generator and calculate the complete matrix diagnostics pipeline:

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# DO NOT MODIFY: Synthetic Strain Sensor Data Generator (Given in assignment)
np.random.seed(42)
n_samples = 5000
n_features = 3

# Generating base stationary multivariate normal noise
base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting a subtle, hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015  # Sensor 1 drift
base_data[4000:, 2] -= 0.010  # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])


# --- MANOVA Verification Engine Implementation ---
def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g chunks and evaluates first-moment homogeneity
    via Wilks' Lambda and Bartlett's Chi-Square asymptotic transformation.
    """
    X = df.to_numpy()
    n, m = X.shape
    chunk_size = n // g_chunks

    # 1. Compute the global mean
    global_mean = np.mean(X, axis=0)

    # Initialize matrices W and B with zeros
    W = np.zeros((m, m))
    B = np.zeros((m, m))

    # 2 & 3. Process chunk-by-chunk variables
    for j in range(g_chunks):
        # Extract the exact slice representing the j-th block
        chunk_data = X[j * chunk_size : (j + 1) * chunk_size, :]
        chunk_mean = np.mean(chunk_data, axis=0)

        # Calculate within-chunk deviation matrix layer
        dev_within = chunk_data - chunk_mean
        W += dev_within.T @ dev_within

        # Calculate between-chunk deviation matrix layer
        dev_between = chunk_mean - global_mean
        B += chunk_size * np.outer(dev_between, dev_between)

    # 4. Compute Wilks' Lambda (Determinant Ratio)
    det_W = np.linalg.det(W)
    det_B_plus_W = np.linalg.det(B + W)
    wilks_lambda = det_W / det_B_plus_W

    # Calculate Bartlett's Chi-Square Asymptotic Transformation
    bartlett_factor = n - 1 - (m + g_chunks) / 2.0
    chi2_calc = -bartlett_factor * np.log(wilks_lambda)

    # Compute degrees of freedom and final asymptotic p-value
    df_degrees = m * (g_chunks - 1)
    p_value = 1.0 - stats.chi2.cdf(chi2_calc, df_degrees)

    return {
        "Wilks_Lambda": float(wilks_lambda),
        "Bartlett_Chi2": float(chi2_calc),
        "p_value": float(p_value),
        "df": int(df_degrees)
    }

# Execute completed tracking engine pipeline
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)
print("--- Execution Results Output ---")
for key, value in results.items():
    print(f"{key}: {value:.6f}" if isinstance(value, float) else f"{key}: {value}")
```

```python
import numpy as np
import pandas as pd

class PCASubspaceOptimizer:
    """
    Subspace Optimization Engine for Real-Time Structural Health Monitoring Layers.
    Manages coordinate projections, variance preservation, and diagnostic metrics tracking.
    """
    def __init__(self, data: pd.DataFrame):
        self.raw_data = data.to_numpy()
        self.n, self.m = self.raw_data.shape

    def evaluate_subspace_limits(self, cutoffs=[1, 2, 3]) -> dict:
        # 1. Standardize and center the incoming realizations
        mean_vector = np.mean(self.raw_data, axis=0)
        std_vector = np.std(self.raw_data, axis=0, ddof=1) # Unbiased sample standard deviation

        X_centered = (self.raw_data - mean_vector) / std_vector

        # 2. Compute unbiased sample covariance matrix S
        S = (X_centered.T @ X_centered) / (self.n - 1)

        # Decompose using np.linalg.eigh
        eigenvalues, eigenvectors = np.linalg.eigh(S)

        # Sort components in descending order (highest to lowest variance energy)
        idx = np.argsort(eigenvalues)[::-1]
        lambda_hat = eigenvalues[idx]
        P_hat = eigenvectors[:, idx]

        # Project full standardized realizations onto orthogonal principal space
        Z = X_centered @ P_hat

        subspace_telemetry = {}

        # 3. Loop through every possible subspace cutoff value k
        for k in cutoffs:
            T2_samples = np.zeros(self.n)
            Q_samples = np.zeros(self.n)

            for i in range(self.n):
                # Hotelling's T2 calculation for sample snapshot i
                T2_samples[i] = np.sum((Z[i, :k] ** 2) / lambda_hat[:k])

                # Residual Q (SPE) calculation for sample snapshot i
                Q_samples[i] = np.sum(Z[i, k:] ** 2)

            # 4. Compute expected value (empirical mean) across all samples
            subspace_telemetry[k] = {
                "E_T2": float(np.mean(T2_samples)),
                "E_Q": float(np.mean(Q_samples)),
                "Sum_of_Expectations": float(np.mean(T2_samples) + np.mean(Q_samples))
            }

        return subspace_telemetry

# --- Simulation & Verification Execution Setup ---
if __name__ == "__main__":
    # Generate 3,000 nominal asset snapshots across m=4 collinear sensors
    np.random.seed(42)
    n_samples = 3000
    n_sensors = 4

    # Generate underlying structural loading components driving the data
    latent_loading_1 = np.random.normal(0, 2.0, size=n_samples)
    latent_loading_2 = np.random.normal(0, 0.8, size=n_samples)

    # Create multi-channel collinear matrix tracking original physical load paths
    sensor_data = np.zeros((n_samples, n_sensors))
    sensor_data[:, 0] = 0.8 * latent_loading_1 + 0.1 * latent_loading_2 + np.random.normal(0, 0.2, n_samples)
    sensor_data[:, 1] = 0.7 * latent_loading_1 + 0.3 * latent_loading_2 + np.random.normal(0, 0.2, n_samples)
    sensor_data[:, 2] = 0.2 * latent_loading_1 + 0.9 * latent_loading_2 + np.random.normal(0, 0.1, n_samples)
    sensor_data[:, 3] = 0.5 * latent_loading_1 + 0.5 * latent_loading_2 + np.random.normal(0, 0.3, n_samples)

    df_sensors = pd.DataFrame(sensor_data, columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4'])

    # Instantiate pipeline and execute optimization loop
    optimizer = PCASubspaceOptimizer(df_sensors)
    metrics = optimizer.evaluate_subspace_limits(cutoffs=[1, 2, 3])

    # Print clean tracking metrics results
    print(f"{'Cutoff (k)':<12}{'E[T²]':<15}{'E[Q]':<15}{'Total (E[T²] + E[Q])':<20}")
    print("-" * 65)
    for k, val in metrics.items():
        print(f"{k:<12}{val['E_T2']:<15.6f}{val['E_Q']:<15.6f}{val['Sum_of_Expectations']:<20.6f}")
```

# **Latent Subspace Decomposition via Factor Analysis (FA)**

## **Part A: Theoretical Fundamentals (Generative Model & Commonalities) Answer**

**Question 1: The Fundamental Equation & Variance Partitioning1.**

 **1.Proof of the Fundamental Equation of Factor Analysis ($R = \Lambda\Lambda^T + \Psi$)**

 We are given the linear generative model $Z_i = \Lambda F_i + \epsilon_i$, where the expected values of the factors and errors are zero ($\mathbb{E}[F_i] = 0$, $\mathbb{E}[\epsilon_i] = 0$).To find the population correlation matrix $R$, we evaluate the expectation of the outer product of the observation vector $Z_i$:$$R = \mathbb{E}[Z_i Z_i^T]$$

 Step 1: Substitute the generative structure$$R = \mathbb{E}[(\Lambda F_i + \epsilon_i)(\Lambda F_i + \epsilon_i)^T]$$

 Step 2: Expand the transpose and matrix productsUsing the property $(A + B)^T = A^T + B^T$ and $(AB)^T = B^T A^T$:$$R = \mathbb{E}[(\Lambda F_i + \epsilon_i)(F_i^T \Lambda^T + \epsilon_i^T)]$$$$R = \mathbb{E}[\Lambda F_i F_i^T \Lambda^T + \Lambda F_i \epsilon_i^T + \epsilon_i F_i^T \Lambda^T + \epsilon_i \epsilon_i^T]$$Step 3: Apply the linear expectation operator$$R = \Lambda \mathbb{E}[F_i F_i^T] \Lambda^T + \Lambda \mathbb{E}[F_i \epsilon_i^T] + \mathbb{E}[\epsilon_i F_i^T] \Lambda^T + \mathbb{E}[\epsilon_i \epsilon_i^T]$$Step 4: Inject system constraints and simplifyBased on our defined constraints:$\mathbb{E}[F_i F_i^T] = I_{k \times k}$ (Factors are standardized and orthogonal)$\mathbb{E}[\epsilon_i F_i^T] = 0$ and $\mathbb{E}[F_i \epsilon_i^T] = 0$ (Latent factors and measurement errors are completely independent)$\mathbb{E}[\epsilon_i \epsilon_i^T] = \Psi$ (Diagonalized measurement noise)Substituting these values yields the final Fundamental Equation:$$R = \Lambda (I_{k \times k}) \Lambda^T + 0 + 0 + \Psi$$$$R = \Lambda\Lambda^T + \Psi$$2.

 **2.Definitions and Physical Interpretations of Communality and Uniqueness**

 If we look at the $j$-th diagonal entry of the equation above, it represents the total standardized variance of sensor $j$, which equals 1 ($R_{jj} = 1$). It is cleanly partitioned into two components: $1 = h_j^2 + \varphi_j^2$.

 **Communality ($h_j^2$):**

 Mathematical Definition: The sum of squared factor loadings for the $j$-th variable: $h_j^2 = \sum_{l=1}^k \lambda_{jl}^2$.

 Physical Interpretation: This is the percentage of a sensor's total variance that is driven by the shared structural dynamics of the asset. It represents the true, underlying physical mechanisms (e.g., operational loads, wind stress) that simultaneously affect multiple parts of the machine.

 **Uniqueness ($\varphi_j^2$):**

 Mathematical Definition: The $j$-th diagonal entry of the noise matrix $\Psi$.

 Physical Interpretation: This is the proportion of variance isolated entirely to that specific sensor. In diagnostics, a high uniqueness value physically maps to localized phenomena such as instrumentation faults, calibration errors, or specific localized noise that does not transmit through the rest of the structural frame.

**Question 2: Orthogonal Rotations (Varimax)1.**

 **1.Raw Principal Component vs. Varimax-Rotated Factor Loading**

 **Raw PCA Loading Vector:** Driven by global variance maximization. The first PC acts as a massive "sweep" axis, attempting to capture as much total variance as possible. This mathematically forces the first loading vector to be highly dense, with almost all sensors loading heavily onto it, blurring distinct physical mechanisms together.

 **Varimax-Rotated FA Vector:** Driven by "Simple Structure." Instead of maximizing global variance, it maximizes the variance of the squared loadings within each column. It conceptually forces loadings toward 1 or 0, ensuring that each factor represents a clean, distinct, and highly isolated physical sub-system or load path.

 **2.Why Orthogonal Rotations Preserve Global Metrics**

  Let $T$ represent an orthogonal rotation matrix where $T T^T = I_{k \times k}$. We apply this rotation to the loading matrix to create a new matrix $\tilde{\Lambda} = \Lambda T$.If we substitute this rotated matrix back into the Fundamental Equation approximation:$$R \approx \tilde{\Lambda}\tilde{\Lambda}^T + \Psi = (\Lambda T)(\Lambda T)^T + \Psi$$$$= \Lambda (T T^T) \Lambda^T + \Psi = \Lambda (I) \Lambda^T + \Psi = \Lambda\Lambda^T + \Psi$$
  
  Conclusion: The total covariance reconstruction $R$ is completely untouched. Because the overall matrix does not change, its diagonal elements—the Communalities ($h_j^2$)—also remain mathematically identical. The rotation merely redistributes the shared variance between the hidden factors for better interpretability, without creating or destroying any total system energy.

**Question 3: Thomson’s Regression Method for Factor Scores1.**

**1.Construction of the Joint Covariance Matrix $\Sigma_{YY}$**

We define the joint partitioned vector $Y_i = \begin{bmatrix} Z_i \\ F_i \end{bmatrix}$. Because both sets of variables are zero-mean, the joint covariance matrix is simply the expectation of their outer product:$$\Sigma_{YY} = \mathbb{E}[Y_i Y_i^T] = \mathbb{E}\left[ \begin{bmatrix} Z_i \\ F_i \end{bmatrix} \begin{bmatrix} Z_i^T & F_i^T \end{bmatrix} \right] = \begin{bmatrix} \mathbb{E}[Z_i Z_i^T] & \mathbb{E}[Z_i F_i^T] \\ \mathbb{E}[F_i Z_i^T] & \mathbb{E}[F_i F_i^T] \end{bmatrix}$$Let's solve for each block:Top-Left: $\mathbb{E}[Z_i Z_i^T] = R$ (by fundamental definition).Bottom-Right: $\mathbb{E}[F_i F_i^T] = I_{k \times k}$ (factors are standardized and orthogonal).Top-Right (Cross-Covariance):$$\mathbb{E}[Z_i F_i^T] = \mathbb{E}[(\Lambda F_i + \epsilon_i) F_i^T] = \Lambda \mathbb{E}[F_i F_i^T] + \mathbb{E}[\epsilon_i F_i^T]$$Applying constraints ($\mathbb{E}[F_i F_i^T] = I$ and $\mathbb{E}[\epsilon_i F_i^T] = 0$):$$\mathbb{E}[Z_i F_i^T] = \Lambda(I) + 0 = \Lambda$$Bottom-Left: By symmetry, $\mathbb{E}[F_i Z_i^T] = \Lambda^T$.Mapping these blocks back into the matrix confirms the exact expansion:$$\Sigma_{YY} = \begin{bmatrix} R & \Lambda \\ \Lambda^T & I_{k \times k} \end{bmatrix}$$

**2.Derivation of Thomson’s Estimator ($f_i$)**

We treat the realization of the hidden factors as a conditional expectation problem, applying the given multivariate normal projection identity:$$\mathbb{E}[X_2 | x_1] = \mu_2 + \Sigma_{21}\Sigma_{11}^{-1}(x_1 - \mu_1)$$

Step 1: Map the partitioned variables from $\Sigma_{YY}$ to the formula:$X_2 = F_i$ (The target variables to estimate)$x_1 = z_i$ (The observed sensor data)$\mu_2 = \mathbb{E}[F_i] = 0$$\mu_1 = \mathbb{E}[Z_i] = 0$$\Sigma_{21} = \Lambda^T$ (Bottom-Left block mapping $F$ against $Z$)$\Sigma_{11} = R$ (Top-Left block mapping $Z$ against $Z$)

Step 2: Substitute and solve:$$f_i = \mathbb{E}[F_i | z_i] = 0 + \Lambda^T R^{-1} (z_i - 0)$$This yields the final rigorous scoring equation used by the diagnostic tracking engine:$$f_i = \Lambda^T R^{-1} z_i$$

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots

class FactorAnalysisEngine:
    def __init__(self, df: pd.DataFrame):
        self.df = df

    def compute_empirical_fa(self, k: int, target_cols=None):
        if target_cols is None:
            target_cols = list(self.df.select_dtypes(include=[np.number]).columns)

        X = self.df[target_cols].dropna().values
        n, m = X.shape

        # --- 1. Dimensionality Constraint Layer ---
        if n <= m:
            raise ValueError(f"Operational snapshot count (n={n}) must be strictly greater than channel count (m={m}).")
        if k >= m:
            raise ValueError(f"Requested latent dimension (k={k}) must be strictly less than feature dimension (m={m}).")

        # --- 2. Z-Score Mapping & Pre-processing Guardrails ---
        mu_hat = np.mean(X, axis=0)
        std_hat = np.std(X, axis=0, ddof=1)

        # Guardrail: Cap zero-variance channels to prevent divide-by-zero
        std_hat[std_hat == 0] = 1e-15

        Z = (X - mu_hat) / std_hat

        # --- 3. Information Decomposition Layer (FA with Varimax) ---
        fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
        fa.fit(Z)

        # Extract rotated parameters
        lambda_matrix = fa.components_.T
        uniqueness = fa.noise_variance_
        communality = np.sum(lambda_matrix**2, axis=1)

        # --- 4. Latent Score Projection (Thomson's MMSE) ---
        F_scores = fa.transform(Z)
        factor_variances = np.var(F_scores, axis=0, ddof=1)

        # --- 5. Performance Diagnostic Telemetry ---
        avg_communality = np.mean(communality) * 100
        avg_uniqueness = np.mean(uniqueness) * 100

        print("\n--- FA System Diagnostic Telemetry ---")
        print(f"Average System Communality: {avg_communality:.2f}%")
        print(f"Average System Uniqueness:  {avg_uniqueness:.2f}%")

        # --- 6. Layout Engineering & Interface Design ---
        factor_labels = [f"Factor {j+1}" for j in range(k)]

        fig = make_subplots(
            rows=2, cols=2,
            horizontal_spacing=0.24,
            vertical_spacing=0.28,
            subplot_titles=(
                "Structural Loadings Matrix |Λ_rotated|",
                "Variance Partitioning (Communality vs Uniqueness)",
                "Sensor Uniqueness Noise Floor (φ²)",
                "Latent Factor Scores Empirical Variance"
            )
        )

        # [Row 1, Col 1]: Structural Loadings Heatmap
        fig.add_trace(go.Heatmap(
            z=np.abs(lambda_matrix), x=factor_labels, y=target_cols,
            colorscale='YlOrRd', showscale=True,
            colorbar=dict(x=-0.15, len=0.38, y=0.78, yanchor="middle", xanchor="right", title="Weight")
        ), row=1, col=1)

        # [Row 1, Col 2]: Horizontal Stacked Bar Chart
        fig.add_trace(go.Bar(
            y=target_cols, x=communality * 100, name="Shared Communality (h²)",
            orientation='h', marker_color='#1f77b4'
        ), row=1, col=2)

        fig.add_trace(go.Bar(
            y=target_cols, x=uniqueness * 100, name="Isolated Uniqueness (φ²)",
            orientation='h', marker_color='#ff7f0e'
        ), row=1, col=2)

        fig.update_layout(barmode='stack')
        fig.update_xaxes(range=[0, 100], title_text="Variance (%)", row=1, col=2)

        # [Row 2, Col 1]: Sensor Noise Floor Line
        fig.add_trace(go.Scatter(
            x=target_cols, y=uniqueness, name="Uniqueness",
            mode='lines+markers', line=dict(color='#d62728', width=2, dash='dashdot'),
            marker=dict(symbol='x', size=8)
        ), row=2, col=1)

        fig.update_xaxes(tickangle=25, row=2, col=1)
        fig.update_yaxes(title_text="Noise Variance", row=2, col=1)

        # [Row 2, Col 2]: Latent Empirical Variance
        fig.add_trace(go.Bar(
            x=factor_labels, y=factor_variances, name="Factor Variance",
            marker=dict(color='#2ca02c', line=dict(color='black', width=0.5))
        ), row=2, col=2)

        fig.update_yaxes(title_text="Empirical Variance", row=2, col=2)

        # Global Canvas Constraints
        fig.update_layout(
            width=1250, height=750, template="plotly_white",
            margin=dict(t=150, b=60, l=140, r=80),
            legend=dict(orientation="h", x=0.5, y=1.02, xanchor="center")
        )

        fig.show()

# ==========================================
# Execution Example
# ==========================================
if __name__ == "__main__":
    # Generate Synthetic Dummy Data for Execution
    np.random.seed(42)
    n_samples = 2500

    f1 = np.random.normal(0, 1, n_samples)
    f2 = np.random.normal(0, 1, n_samples)

    s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3, n_samples)
    s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)
    s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)
    s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)

    df_asset = pd.DataFrame({'Sensor_1': s1, 'Sensor_2': s2, 'Sensor_3': s3, 'Sensor_4': s4})

    # Run Pipeline
    engine = FactorAnalysisEngine(df_asset)
    engine.compute_empirical_fa(k=2)
```

```python
import numpy as np
import pandas as pd
from typing import Optional, Sequence, Dict, Any
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# =====================================================================
# COMBINED SUBSPACE ENGINES (PCA & FA PIPELINES)
# =====================================================================

class DualSubspaceDiagnosticsEngine:
    def __init__(self, df: pd.DataFrame):
        self.df = df

    def compute_empirical_pca(self, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the geometric de-correlation framework via PCA.
        Decomposes the unbiased empirical covariance matrix S into its orthogonal
        basis P_hat and renders a clean 2x3 optimization and feature contribution dashboard.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        if n <= m:
            raise ValueError("Sample size n must be larger than feature dimensions m to isolate principal components.")

        # 1. Centering
        mu_hat = np.mean(X, axis=0)
        X_centered = X - mu_hat

        # 2. Compute Unbiased Sample Covariance Matrix S
        S_matrix = np.cov(X, rowvar=False, ddof=1)

        # 3. Spectral Decomposition
        eigenvalues, eigenvectors = np.linalg.eigh(S_matrix)
        idx = np.argsort(eigenvalues)[::-1]
        lambda_hat = eigenvalues[idx]
        P_hat = eigenvectors[:, idx]
        lambda_hat = np.clip(lambda_hat, a_min=1e-15, a_max=None)

        # 4. Variance Information Calculations
        total_variance = np.sum(lambda_hat)
        if total_variance == 0:
            raise ValueError("Total variance is zero.")

        explained_variance_ratio = lambda_hat / total_variance
        cumulative_variance_ratio = np.cumsum(explained_variance_ratio)
        unexplained_variance_ratio = 1.0 - cumulative_variance_ratio

        # 5. Map Realizations to Principal Component Scores
        Z_scores = np.dot(X_centered, P_hat)
        S_Z = np.cov(Z_scores, rowvar=False, ddof=1)

        # 6. Evaluate Statistical Distance Metrics (T^2 and Q)
        k_range = np.arange(1, m)
        mean_T2_vs_k = []
        mean_Q_vs_k = []
        T2_matrix = np.zeros((n, len(k_range)))
        Q_matrix = np.zeros((n, len(k_range)))

        for idx_k, k in enumerate(k_range):
            Z_k = Z_scores[:, :k]
            lambda_k = lambda_hat[:k]
            T2_samples = np.sum((Z_k ** 2) / lambda_k, axis=1)
            T2_matrix[:, idx_k] = T2_samples
            mean_T2_vs_k.append(np.mean(T2_samples))

            Z_residual = Z_scores[:, k:]
            Q_samples = np.sum(Z_residual ** 2, axis=1)
            Q_matrix[:, idx_k] = Q_samples
            mean_Q_vs_k.append(np.mean(Q_samples))

        print(f"\n--- Operationalizing Geometric De-correlation Layer (PCA) ---")
        print(f"Decomposing structural space of {m} features using {n} samples.")
        print(f"Total System Variance (Trace[S]): {total_variance:.4f}")

        # 7. Enhanced 2x3 Diagnostic Dashboard
        if show_plot:
            pc_labels = [f"PC {i+1}" for i in range(m)]
            k_labels = [f"k={k}" for k in k_range]
            sensor_labels = target_cols

            fig = make_subplots(
                rows=2, cols=3,
                horizontal_spacing=0.18,
                vertical_spacing=0.28,
                subplot_titles=(
                    "Feature Loading Matrix |P_hat|",
                    "Component Values (Eigenvalues λ)",
                    "Information Profile (Explained Var.)",
                    "Residual Space (Unexplained Var.)",
                    "Mean Hotelling's T² vs Subspace Size k",
                    "Mean Q Statistic (SPE) vs Subspace Size k"
                )
            )

            # --- ROW 1, COL 1: PCA Loading Heatmap (Left-anchored Colorbar) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(P_hat),
                    x=pc_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Loading Weight",
                        x=-0.12,          # Perfectly isolates colorbar to the left side
                        len=0.38,         # Fits height bounds of row 1
                        y=0.78,           # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right",
                        titleside="top"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Principal Axes", row=1, col=1)

            # --- ROW 1, COL 2: Absolute Eigenvalues ---
            fig.add_trace(go.Bar(x=pc_labels, y=lambda_hat, name="Eigenvalue (λ_j)", marker=dict(color='#1f77b4', line=dict(color='black', width=0.5)), legendgroup="eigen"), row=1, col=2)
            fig.update_yaxes(title_text="Variance Magnitude", row=1, col=2)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=2)

            # --- ROW 1, COL 3: Variance Profiles ---
            fig.add_trace(go.Bar(x=pc_labels, y=explained_variance_ratio * 100, name="Marginal Explained", marker=dict(color='#ff7f0e', opacity=0.75), legendgroup="expl"), row=1, col=3)
            fig.add_trace(go.Scatter(x=pc_labels, y=cumulative_variance_ratio * 100, mode='lines+markers', name='Cumulative Captured', line=dict(color='#d62728', width=2.5, dash='dash'), legendgroup="expl"), row=1, col=3)
            fig.update_yaxes(title_text="Captured Structure (%)", range=[-2, 105], row=1, col=3)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=3)

            # --- ROW 2, COL 1: Remaining Noise ---
            fig.add_trace(go.Bar(x=pc_labels, y=unexplained_variance_ratio * 100, name="Remaining Noise", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="noise"), row=2, col=1)
            fig.update_yaxes(title_text="Excluded Info (%)", range=[-2, 105], row=2, col=1)
            fig.update_xaxes(title_text="Principal Axes", row=2, col=1)

            # --- ROW 2, COL 2: Hotelling's T^2 ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_T2_vs_k, mode='lines+markers', name='Mean T²', line=dict(color='#9467bd', width=2.5), marker=dict(size=6, symbol='diamond'), legendgroup="t2"), row=2, col=2)
            fig.update_yaxes(title_text="Average T² Metric", row=2, col=2)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=2)

            # --- ROW 2, COL 3: SPE / Q Statistic ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_Q_vs_k, mode='lines+markers', name='Mean Q (SPE)', line=dict(color='#e377c2', width=2.5), marker=dict(size=6, symbol='square'), legendgroup="q_stat"), row=2, col=3)
            fig.update_yaxes(title_text="Average Residual Energy", row=2, col=3)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=3)

            # Global Unified Layout Configurations
            fig.update_layout(
                title=dict(text="Principal Component Analysis (PCA) Optimization & Feature Loading Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5
                ),
                margin=dict(t=150, b=60, l=140, r=80),
                height=750,
                width=1250
            )
            fig.show()

        return {
            "mean_vector": mu_hat, "covariance_matrix_S": S_matrix, "eigenvalues_lambda": lambda_hat,
            "eigenvectors_P": P_hat, "explained_variance_ratio": explained_variance_ratio,
            "cumulative_variance_ratio": cumulative_variance_ratio, "unexplained_variance_ratio": unexplained_variance_ratio,
            "transformed_scores_Z": Z_scores, "score_covariance_diagonal": np.diag(S_Z),
            "features": target_cols, "k_values": k_range, "T2_matrix_vs_k": T2_matrix, "Q_matrix_vs_k": Q_matrix,
            "mean_T2_profile": np.array(mean_T2_vs_k), "mean_Q_profile": np.array(mean_Q_vs_k)
        }

    def compute_empirical_fa(self, k: int, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the Factor Analysis latent subspace framework with a clean 2x2 grid.
        Anchors the heatmap colorbar perfectly to the right edge of the first subplot.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        # 1. Standardize realizations to construct Z space
        mu_hat = np.mean(X, axis=0)
        std_hat = np.std(X, axis=0, ddof=1)
        std_hat[std_hat == 0] = 1e-15
        Z = (X - mu_hat) / std_hat
        R_matrix = np.corrcoef(X, rowvar=False)

        # 2. Fit Factor Analysis Model (Varimax Rotation)
        fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
        fa.fit(Z)
        lambda_matrix = fa.components_.T
        psi_diagonal = fa.noise_variance_

        # 3. Calculate Communality (h²) and Uniqueness (φ²) vectors
        communality = np.sum(lambda_matrix**2, axis=1)
        uniqueness = psi_diagonal

        # 4. Map realizations using Thomson's Regression Method
        F_scores = fa.transform(Z)

        print(f"\n--- Operationalizing Latent Subspace Layer (Factor Analysis) ---")
        print(f"Decomposing structural space of {m} sensors into {k} hidden physical factors.")
        print(f"Average System Communality (Shared Subspace Energy): {np.mean(communality)*100:.2f}%")
        print(f"Average System Uniqueness (Localized Noise Floor): {np.mean(uniqueness)*100:.2f}%")

        # 5. Clean 2x2 Dashboard
        if show_plot:
            sensor_labels = target_cols
            factor_labels = [f"Factor {j+1}" for j in range(k)]

            fig = make_subplots(
                rows=2, cols=2, horizontal_spacing=0.24, vertical_spacing=0.28,
                subplot_titles=(
                    "Structural Loadings Matrix |λ_(j,r)|",
                    "Variance Partitioning (Communality vs Uniqueness)",
                    "Sensor Uniqueness Noise Floor (φ²)",
                    "Latent Factor Scores Empirical Variance"
                )
            )

            # --- Subplot 1: Factor Loadings Heatmap (Colorbar repositioned to x=0.42) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(lambda_matrix),
                    x=factor_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Sensitivity",
                        x=-0.1,          # Shifts colorbar out of the far right to center-left
                        len=0.38,         # Scales down length to match row 1 height bounds
                        y=0.78,          # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Latent Structures", row=1, col=1)

            # --- Subplot 2: Variance Partitioning ---
            fig.add_trace(go.Bar(y=sensor_labels, x=communality * 100, name="Communality (h² - Shared Structure)", orientation='h', marker=dict(color='#1f77b4'), legendgroup="communality"), row=1, col=2)
            fig.add_trace(go.Bar(y=sensor_labels, x=uniqueness * 100, name="Uniqueness (φ² - Channel Noise)", orientation='h', marker=dict(color='#ff7f0e'), legendgroup="uniqueness_bar"), row=1, col=2)
            fig.update_layout(barmode='stack')
            fig.update_xaxes(title_text="Variance Allocation (%)", range=[0, 100], row=1, col=2)

            # --- Subplot 3: Uniqueness Line Profile ---
            fig.add_trace(go.Scatter(x=sensor_labels, y=uniqueness, mode='lines+markers', name='Uniqueness Profile (φ²)', line=dict(color='#d62728', width=2, dash='dot'), marker=dict(size=8, symbol='x'), legendgroup="uniqueness_line"), row=2, col=1)
            fig.update_yaxes(range=[-0.05, 1.05], row=2, col=1)
            fig.update_xaxes(title_text="Monitored Channels", tickangle=25, row=2, col=1)

            # --- Subplot 4: Latent Factor Scores Empirical Variance ---
            factor_variances = np.var(F_scores, axis=0, ddof=1)
            fig.add_trace(go.Bar(x=factor_labels, y=factor_variances, name="Factor Empirical Variance", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="variance"), row=2, col=2)
            fig.update_yaxes(title_text="Variance Level", row=2, col=2)
            fig.update_xaxes(title_text="Latent Vectors", row=2, col=2)

            # Global Layout Updates
            fig.update_layout(
                title=dict(text="Factor Analysis (FA) Latent Subspace Diagnostics Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5,
                    traceorder="normal",
                    itemwidth=40,
                    itemsizing="constant"
                ),
                margin=dict(t=100, b=60, l=80, r=80),
                height=750,
                width=1100
            )
            fig.show()

        return {
            "mean_vector_mu": mu_hat, "std_vector_D": std_hat, "correlation_matrix_R": R_matrix,
            "factor_loadings_lambda": lambda_matrix, "uniqueness_psi": uniqueness,
            "communality_h2": communality, "latent_factor_scores_F": F_scores, "sensors": target_cols
        }

# =====================================================================
# DATA GENERATOR & PIPELINE EXECUTION
# =====================================================================
if __name__ == "__main__":
    # Set execution seed for perfect reproducible properties
    np.random.seed(42)
    n_samples = 2500

    # 1. Generate underlying, independent true physical driving forces (Latent Factors)
    f1 = np.random.normal(0, 1, n_samples)  # Primary structural mode/load
    f2 = np.random.normal(0, 1, n_samples)  # Secondary operational mode

    # 2. Construct the Observable Multi-Sensor Array with targeted noise profiles
    # Sensors 1 and 2 couple strongly to the primary process (f1)
    s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3, n_samples)
    s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)

    # Sensor 3 couples strongly to the secondary process (f2)
    s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)

    # Sensor 4 suffers an extreme electrical/instrumentation fault (massive localized white noise)
    s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)

    # 3. Compile into the final evaluation DataFrame
    df_asset = pd.DataFrame(
        data=np.vstack([s1, s2, s3, s4]).T,
        columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
    )
    # Instantiate Engine
    engine = DualSubspaceDiagnosticsEngine(df_asset)

    # Run PCA Engine (1x5 Dashboard)
    pca_metrics = engine.compute_empirical_pca(show_plot=True)

    # Run FA Engine (2x2 Dashboard, Retaining k=2 Common Physical Factors)
    fa_metrics = engine.compute_empirical_fa(k=2, show_plot=True)
```

## **PART B ANSWERS**

### answer for Question 1: The Total Variance Illusion

* **Part 1 (Physical Nature of Sensor 4):** A Uniqueness value ($\varphi^2$) approaching $100\%$ indicates that `Sensor_4` shares zero covariance with the rest of the array. It is completely decoupled from the real physical processes ($f_1, f_2$) driving the plant asset. Its data profile consists entirely of isolated white noise or severe electrical fault energy.
* **Part 2 (The PCA Loading Contamination):** Looking at the **Feature Loading Matrix** in the PCA dashboard, we see a striking anomaly: `PC 1` maps almost exclusively to a single variable—`Sensor_4` (indicated by the deep red block with a loading value near $1.0$). Because `Sensor_4` has a massive, unmitigated noise variance scale ($\sigma^2 \approx 2.0$), the PCA engine maximizes global variance by dedicating its entire primary axis (`PC 1`) to tracking this single noisy sensor. It completely isolates `Sensor_1`, `Sensor_2`, and `Sensor_3` to a loading of $0$ on `PC 1`.
* **Part 3 (Engineering Risk):** This presents a significant operational hazard. If you use standard PCA for monitoring, a single blown or short-circuited instrument will hijack your most critical monitoring axis (`PC 1`), inflating its eigenvalue to $\lambda_1 \approx 2.0$. This masks the true common underlying structural dynamics of the asset and renders your primary control chart useless for catching actual systemic failures.

### Answer for Question 2: Decoupling Structural Loading via Rotation

* **Part 1 (Mathematical Jumble vs. Simple Structure):** Because PCA is bound to strict, orthogonal variance maximization, look at what happens to the remaining signals on `PC 2` and `PC 4`. `PC 2` and `PC 4` exhibit high loading values across *both* `Sensor_1` and `Sensor_2` simultaneously, while also bleeding into `Sensor_3`. This cross-loading represents a mathematically forced mixture of signals. Factor Analysis applies a **Varimax rotation** to eliminate this. By maximizing loading variance within columns, it completely isolates the signals: `Sensor_1` and `Sensor_2` group cleanly into `Factor 1`, while `Sensor_3` shifts entirely into `Factor 2`.
* **Part 2 (Operational Interpretability):** The PCA loading matrix forces a complex interpretation. If a diagnostic alarm trips on `PC 2`, an engineer has to figure out a messy combination of Sensors 1, 2, and 3. With the rotated FA framework, the layout is intuitive. `Factor 1` represents the isolated physical process causing redundancy in `Sensor_1` and `Sensor_2`, while `Factor 2` maps directly to the independent process monitored by `Sensor_3`. Troubleshooting root causes becomes instantaneous.


### Answers for Question 3: Determining Subspace Truncation ($k$)

* **Part 1 (Curve Observations):** Moving from $k=1$ to $k=2$, the Mean $Q$ Statistic (SPE) plunges from $2.4$ down to $0.9$. When transitioning from $k=2$ to $k=3$, the curve flattens out into an explicit "elbow", decreasing at a much slower, linear rate down toward $0.1$.
* **Part 2 (The True Intrinsic Dimension):** Cross-referencing the $Q$ curve with the new loading matrix makes the selection clear. `PC 1` tracks the isolated noise of `Sensor_4`. `PC 2` captures the true, shared variance of the working sensors (`Sensor_1`, `Sensor_2`, `Sensor_3`). The sharp drop in the $Q$ statistic at $k=2$ tells us that once these two distinct behaviors are accounted for, the structured information is fully exhausted. Thus, $k=2$ represents the true hidden physical dimensionality of the asset.
* **Part 3 (The Danger of Over-fitting to $k=3$):** Look at `PC 3` in the loading matrix—it maps exclusively to the leftover residual variance of `Sensor_3`. Setting your truncation limit to $k=3$ pulls this residual space into your "clean" model subspace. This overfits the system, causing your Hotelling's $T^2$ control limits to trigger frequent false alarms due to minor, normal baseline variations.



### Answers for Question 4: Operational Trade-offs in System Health Monitoring

* **The PCA Strategy:** Tracking the main principal subspace via $T^2$ alongside the residual space via the $Q$ statistic.
* *Vulnerability:* This approach is highly vulnerable to sensor degradation. As shown in the PCA loading matrix, if a single channel (`Sensor_4`) fails high, it structurally shifts the orientation of your principal axes, dragging the calculations for the remaining healthy channels along with it.


* **The FA Strategy:** Directly monitoring the rotated Latent Factor Scores.
* *Robustness:* This approach is highly robust. Because Factor Analysis explicitly separates localized noise via the diagonal uniqueness vector ($\Psi$), any single-channel sensor failure is safely absorbed by that channel's uniqueness profile ($\psi_4^2 \rightarrow 1$). The underlying latent factors (`Factor 1` and `Factor 2`) remain clean, stable, and uncorrupted, allowing the rest of your health monitoring network to function normally.
